```python
#Etapa 1:
#Ingestão e Leitura dos Dados

#importar a biblioteca Pandas
import pandas as pd

#criar as variaveis de acordo com as planilhas
dados_logistica = "/content/entregas_orquideas.csv"
dados_agricolas = "/content/estufas_qualidade.csv"
dados_telemetria = "/content/telemetria_iot.csv"

#Criar os dataframe de acordo com as variaveis
df_dados_logistica = pd.read_csv(dados_logistica)
df_dados_agricolas = pd.read_csv(dados_agricolas)
df_dados_telemetria = pd.read_csv(dados_telemetria)

#Exibição das planilhas
display(df_dados_logistica.head(1000))
display(df_dados_agricolas.head(1000))
display(df_dados_telemetria.head(1000))
```

```python
#Etapa 2: Classificação com Estruturas de Repetição
# Monitoramento de temperatura

#Criação de lista para verificar temperatura
list_status_temperatura = []

list_temp = df_dados_telemetria["Temperatura_C"]. tolist()

#Criar loop com for, para conferir a coluna "dados temperadura"
for t in list_temp:
  #Criar parametro para mostrar mensagem de ALERTA ou OK, de acordo com a temperatura exibida
  if t >=28:
    list_status_temperatura.append("ALERTA")

  else:
    list_status_temperatura.append("OK")

#Exibir nova coluna com a informação do if e else (ALERTA ou OK)
df_dados_telemetria["dados_temperatura"] = list_status_temperatura

display(df_dados_telemetria)
```

```python
#Etapa 3: Estatística Básica e Diagnóstico
#01 - Quantidade total de orquídeas transportadas

df_diagnostico_geral = df_dados_logistica

display(df_diagnostico_geral)

print(f"A quantidade total de orquídeas transportadas é: {df_diagnostico_geral["Qtd_Orquideas"].sum()}")

print(f"A quantidade total de orquídeas perdidas: {df_diagnostico_geral["Qtd_Orquideas"].sum()}")
```

```python
#02 - Quantidade total de orquídeas perdidas no transporte

df_perdas = df_dados_logistica[df_dados_logistica["Status_Entrega"] == "Cancelado"]

print(f"A quantidade total de orquídeas perdidas: {df_perdas["Qtd_Orquideas"].sum()}")
```

```python
#03 - Impacto Regional, estado que vai sofreu com cancelamentos de entrega

df_estados = df_dados_logistica[df_dados_logistica["Status_Entrega"] == "Cancelado"]

print(f"O estado que mais sofreu com os cancelamentos:{df_estados["UF_Destino"].mode()}")
```

```python
#04 - Qual foi a temperatura máxima e a temperatura média?
df_max_temperatura = df_dados_telemetria

print(f" A temperatura máxima foi: {df_max_temperatura["Temperatura_C"].max()}")

print(f" A temperatura média foi: {df_max_temperatura["Temperatura_C"].median()}")
```
